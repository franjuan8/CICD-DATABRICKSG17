```python
%python

storageLocation = dbutils.widgets.get("storageLocation")
catalog = dbutils.widgets.get("catalog")

rutaBronze = f"abfss://bronze@{storageLocation}.dfs.core.windows.net"
rutaSilver = f"abfss://silver@{storageLocation}.dfs.core.windows.net"
rutaGolden = f"abfss://golden@{storageLocation}.dfs.core.windows.net"
```

### Eliminacion tablas Bronze

```python
%sql
-- DROP TABLES BRONZE
DROP TABLE IF EXISTS ${catalog}.bronze.bbdd_dim_aerolineas;
DROP TABLE IF EXISTS ${catalog}.bronze.bbdd_dim_aeropuertos;
DROP TABLE IF EXISTS ${catalog}.bronze.bbdd_dim_aeronaves;
DROP TABLE IF EXISTS ${catalog}.bronze.bbdd_dim_rutas;
DROP TABLE IF EXISTS ${catalog}.bronze.bbdd_fact_vuelos;
DROP TABLE IF EXISTS ${catalog}.bronze.bbdd_fact_pasajeros;
DROP TABLE IF EXISTS ${catalog}.bronze.bbdd_fact_combustible;
DROP TABLE IF EXISTS ${catalog}.bronze.bbdd_fact_ventas_boletos;
DROP TABLE IF EXISTS ${catalog}.bronze.bbdd_fact_equipajes;
```

```python
%python
## REMOVE DATA (Bronze)
dbutils.fs.rm(f"{rutaBronze}/bbdd_dim_aerolineas", True)
dbutils.fs.rm(f"{rutaBronze}/bbdd_dim_aeropuertos", True)
dbutils.fs.rm(f"{rutaBronze}/bbdd_dim_aeronaves", True)
dbutils.fs.rm(f"{rutaBronze}/bbdd_dim_rutas", True)
dbutils.fs.rm(f"{rutaBronze}/bbdd_fact_vuelos", True)
dbutils.fs.rm(f"{rutaBronze}/bbdd_fact_pasajeros", True)
dbutils.fs.rm(f"{rutaBronze}/bbdd_fact_combustible", True)
dbutils.fs.rm(f"{rutaBronze}/bbdd_fact_ventas_boletos", True)
dbutils.fs.rm(f"{rutaBronze}/bbdd_fact_equipajes", True)
```
